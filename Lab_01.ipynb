**Task:06** Implementation of multithreading using the threading module to execute multiple threads concurrently and perform parallel operations.

```python
import threading
#fucntion 01
def intro():
  print("Hi! I am GPT-6-Astra. How I can help you")

#function 02
def add(a,b):
  print(a+b)


#function 03
def info():
  print("My name is Abdul-Mateen\nI am 20 years old\nCurrently I am studying in BSCS")

t1=threading.Thread(target=intro)
t2=threading.Thread(target=add, args=(3, 10))
t3=threading.Thread(target=info)
t1.start()
t2.start()
t3.start()
```

**Task:07** Implementation of multiprocessing using the multiprocessing module to execute tasks

```python
import multiprocessing
#fucntion 01
def intro():
  print("Hi! I am GPT-6-Astra. How I can help you")

#function 02
def add(a,b):
  print(a+b)

#function 03
def info():
  print("My name is Abdul-Mateen\nI am 20 years old\nCurrently I am studying in BSCS")

p1=multiprocessing.Process(target=intro)
p2=multiprocessing.Process(target=add, args=(11,12))
p3=multiprocessing.Process(target=info)

p1.start()
p2.start()
p3.start()
```
